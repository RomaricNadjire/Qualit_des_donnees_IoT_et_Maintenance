### 4. Identification des créneaux horaires d'indisponibilité.

Nous allons tout d'abord extraire l'heure dans une nouvelle collonne que nous nommons `heure`. Cette collonne nous servira à regrouper les enregistrements par heure afin d'identifier les heures qui n'ont pas d'inactivités

```python
non_ok_sol_df.loc[:, "heure"] = non_ok_sol_df["date_heure"].dt.time
non_ok_sol_df.head()
```

Comptage des enregistrements par heure

```python
indispo_hours = (
    non_ok_sol_df
    .groupby(["heure", "id_capteur"])["id_capteur"]
    .count()
    .reset_index(name="nb_incidents")
)
print(indispo_hours)
```

```python
pivot_table = indispo_hours.pivot(
    index="id_capteur",
    columns="heure",
    values="nb_incidents"
)

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12,8))

sns.heatmap(pivot_table, cmap="Reds", annot=True)

plt.title("Créneaux d'indisponibilité par capteur")
plt.xlabel("Heure")
plt.ylabel("Capteur")

plt.show()
```

La heatmap des créneaux d’indisponibilité met en évidence des différences importantes entre les capteurs et les zones. Les capteurs de la zone D sont les plus affectés, avec des fréquences d’incidents particulièrement élevées sur l’ensemble de la période étudiée. Le capteur `SD02` atteint par exemple jusqu’à **33 incidents** autour de 13h, tandis que `SD04` présente également des valeurs très élevées avec un maximum de **31 incidents** vers 03h. Les capteurs `SD03`, `SD05` et `SD06` affichent eux aussi régulièrement des fréquences comprises entre **20 et 30 incidents** selon les heures.

Les capteurs de la zone C montrent un niveau intermédiaire d’indisponibilité. Plusieurs créneaux atteignent entre **15 et 22 incidents**, notamment pour les capteurs `SC02`, `SC05`, `SC06` et `SC07`. La zone B présente des fréquences plus modérées, généralement comprises entre **5 et 17 incidents**, bien que certains pics soient observés, comme `SB11` avec **21 incidents** vers 10h ou `SB08` avec **21 incidents** à 00h.

À l’inverse, les capteurs de la zone A apparaissent comme les plus stables. Les fréquences d’incidents y restent faibles, généralement entre **1 et 7 incidents** selon les heures, avec quelques valeurs légèrement plus élevées comme `SA08` atteignant **9 incidents** vers 06h.

La présence de fortes valeurs sur certaines heures ne signifie que des états anormaux ont été observés de manière répétée à ces heures au cours de la période étudiée, c'est le nombre de jour ou le capteur ne fonctionnait pas correctement acette heure..

Globalement, cette analyse montre que les zones **C** et surtout **D** nécessitent une attention prioritaire en matière de maintenance, car leurs capteurs présentent des anomalies plus fréquentes et plus récurrentes.

### 5. Proposer une stratégie de redondance
Étant donné que les zones **C** et **D** présentent le plus grand nombre d’incidents et les fréquences d’indisponibilité les plus élevées, elles devraient être considérées comme prioritaires pour le déploiement de mécanismes de redondance.

Une première approche consiste à installer des capteurs supplémentaires dans les zones les plus critiques. Par exemple, les capteurs `SD02`, `SD03`, `SD04`, `SD05` et `SD06` pourraient être doublés par des capteurs de secours placés à proximité afin d’assurer la continuité de la collecte des données en cas de panne.

Il serait également pertinent de mettre en place une redondance logique des mesures. Dans ce cas, les données des capteurs voisins pourraient être utilisées pour estimer temporairement les valeurs manquantes lorsqu’un capteur devient indisponible. Cette méthode permettrait de réduire l’impact des dysfonctionnements sur les analyses.

Une autre stratégie consiste à configurer un système d’alerte automatique capable de détecter une augmentation anormale des incidents sur certains capteurs ou durant certaines plages horaires. Cela permettrait d’intervenir rapidement avant qu’une panne complète ne survienne.

Enfin, un plan de maintenance préventive ciblé devrait être appliqué en priorité aux capteurs les plus instables, notamment ceux de la zone D, afin de réduire la fréquence des indisponibilités et améliorer la fiabilité globale du réseau IoT.

## PARTIE 2 – Analyse de la qualité des données GPS
Objectif : Évaluer la qualité des coordonnées GPS et leurs impacts sur les analyses spatiales.

### 6. Identification des mesures avec precision_gps > 4
Nous allons commencer par filtrer les données pour n'avoir que les enregistrement dont la precisions gps est au dessus de **4**.

```python
precision_4_sol_df = sol_df[sol_df['precision_gps'] > 4].copy()
print(f"Nombre de lignes avec une precision GPS > 4 : {precision_4_sol_df.shape[0]}")
```

Nous allons en suite passer a l'identification des capteur concernés en nous servant du recultat que nous venons d'obtenir.

```python
capteurs_precision_4 = (
    precision_4_sol_df
    .groupby("id_capteur")
    .agg(
        nb_mesures=("id_capteur", "size"),
        zone=("zone", "first")
    )
    .reset_index()
)

capteurs_precision_4
```

Les capteurs présentant des mesures avec une précision GPS supérieure à 4 sont au nombre de 17. 

Les capteurs les plus concernés sont principalement ceux de la série SD (`zone D`), notamment SD05 (684 mesures), SD03 (653 mesures) et SD02 (651 mesures). Les capteurs de la série SC (`Zone C`) présentent également un nombre important d'enregistrements concernés, avec des valeurs comprises entre 296 et 333 mesures.

En revanche, les capteurs de la série SB (`zone B`) sont nettement moins affectés, avec un nombre de mesures variant entre 54 et 78.

Cette répartition suggère que les anomalies de précision GPS sont davantage concentrées sur certains groupes de capteurs, en particulier ceux des zone **D** et **C**. Ce résultat est cohérent avec l'analyse précédente des capteurs défectueux, qui montrait déjà une plus forte concentration des anomalies dans les zones D et C, et dans une moindre mesure dans la zone B.

### 7. Visualisation des capteurs concernés sur une carte

Les données contenant les coordonnées géographiques (latitude et longitude), il est possible de visualiser la position des capteurs concernés sur une carte afin d'observer leur répartition spatiale.

```python
# Position moyenne de chaque capteur
all_capteurs = (
    sol_df
    .groupby("id_capteur")[["latitude", "longitude"]]
    .mean()
    .reset_index()
)

# Qualité moyenne GPS par capteur (au lieu des incidents)
gps_precision = (
    sol_df
    .groupby("id_capteur")["precision_gps"]
    .mean()
    .reset_index(name="precision_moyenne")
)

# Fusion
capteurs_map = all_capteurs.merge(
    gps_precision,
    on="id_capteur",
    how="left"
)

# Extraction de la zone
capteurs_map["zone"] = capteurs_map["id_capteur"].str[1]

# Couleurs par zone
colors = {
    "B": "blue",
    "C": "orange",
    "D": "red"
}

# Centre de la carte
centre_lat = capteurs_map["latitude"].mean()
centre_lon = capteurs_map["longitude"].mean()

# Normalisation (éviter division par 0)
max_prec = capteurs_map["precision_moyenne"].max()
if max_prec == 0:
    max_prec = 1

# Création de la carte
m = folium.Map(
    location=[centre_lat, centre_lon],
    zoom_start=12
)

for _, row in capteurs_map.iterrows():

    color = colors.get(row["zone"], "gray")

    # Taille basée sur la précision GPS
    # (plus la valeur est grande → plus le point est visible)
    radius = 3 + (row["precision_moyenne"] / max_prec) * 10

    folium.CircleMarker(
        location=[row["latitude"], row["longitude"]],
        radius=radius,
        color=color,
        weight=2,
        fill=True,
        fill_color=color,
        fill_opacity=0.5,
        popup=f"""
        Capteur : {row['id_capteur']}<br>
        Zone : {row['zone']}<br>
        Précision GPS moyenne : {row['precision_moyenne']:.2f}
        """,
        tooltip=row["id_capteur"]
    ).add_to(m)

# Ajustement automatique du cadrage
m.fit_bounds([
    [capteurs_map["latitude"].min(), capteurs_map["longitude"].min()],
    [capteurs_map["latitude"].max(), capteurs_map["longitude"].max()]
])

m
```

Les capteurs sont positionnés sur la carte en fonction de leur localisation géographique, et leur couleur dépend de leur zone d’appartenance :

- Zone B --> bleu
- Zone C --> orange
- Zone D --> rouge.

La taille des marqueurs des capteurs est proportionnelle à leur niveau de précision GPS.

### 8. Calcul de la proportion de mesures GPS imprécises par zone

```python
# Total de mesures par zone
total_zone = (
    sol_df
    .groupby("zone")
    .size()
    .reset_index(name="nb_total")
)

# Mesures imprécises (precision_gps > 4)
imprecises_zone = (
    precision_4_sol_df
    .groupby("zone")
    .size()
    .reset_index(name="nb_imprecises")
)

# Fusion
zone_stats = total_zone.merge(
    imprecises_zone,
    on="zone",
    how="left"
)

# Remplacer les zones sans erreurs
zone_stats["nb_imprecises"] = zone_stats["nb_imprecises"].fillna(0)

# Proportion
zone_stats["proportion_imprecise (%)"] = (
    zone_stats["nb_imprecises"] / zone_stats["nb_total"]
) * 100

zone_stats
```

On observe une forte hétérogénéité de la qualité des mesures GPS selon les zones.

- Zone A : aucune mesure imprécise détectée (0%), ce qui indique un environnement très stable ou un bon fonctionnement des capteurs.
- Zone B : faible taux d’imprécision (~2.39%), suggérant une qualité GPS globalement correcte.
- Zone C : dégradation notable de la qualité des signaux avec environ 14.3% de mesures imprécises.
- Zone D : zone la plus problématique, avec plus de 32% de mesures imprécises, suggère des conditions fortement défavorables à la précision GPS.

### 9. Discussion des conséquences de ces imprécisions
Les imprécisions GPS dégradent directement la qualité et la fiabilité des données.

- Localisation faussée : les capteurs peuvent être mal positionnés sur la carte
- Analyses biaisées : comparaison entre zones moins fiable
- Mauvaises décisions : erreurs possibles dans la maintenance ou le suivi du système
- Perte de confiance globale : surtout dans les zones fortement touchées (ex : Zone D)

### 11. Identification des outliers via la méthode IQR

```python
def detect_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]

    return outliers, lower_bound, upper_bound
```

```python
outliers_pluie, low_p, up_p = detect_outliers_iqr(meteo_df, "pluviometrie")
print("Pluviométrie - bornes :", low_p, up_p)
print("Nombre d'outliers :", len(outliers_pluie))

outliers_vent, low_v, up_v = detect_outliers_iqr(meteo_df, "vitesse_vent")
print("Vitesse du vent - bornes :", low_v, up_v)
print("Nombre d'outliers :", len(outliers_vent))
```

```python
capteurs_outliers_pluie = (
    outliers_pluie["id_capteur"]
    .value_counts()
    .reset_index()
)

capteurs_outliers_pluie.columns = ["id_capteur", "nb_outliers_pluie"]

capteurs_outliers_pluie
```
